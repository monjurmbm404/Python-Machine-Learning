Cell 6 — Create a Pandas DataFrame

```python
students = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 22],
    "Score": [85, 92, 88]
})

students
```

Cell 7 — Display Basic Statistics

```python
students.describe()
```

Cell 8 — Simple Line Plo

```python
plt.figure(figsize=(6,4))

plt.plot(
    students["Name"],
    students["Score"],
    marker="o"
)

plt.title("Student Scores")
plt.xlabel("Student")
plt.ylabel("Score")

plt.grid(True)

plt.show()
```

Cell 9 — Seaborn Bar Plot

```python
sns.barplot(
    data=students,
    x="Name",
    y="Score"
)

plt.title("Student Scores")
plt.show()
```

Cell 10 — Simple Machine Learning Example

```python
from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4], [5]]
y = [2, 4, 6, 8, 10]

model = LinearRegression()

model.fit(X, y)

prediction = model.predict([[6]])

print("Prediction for 6:", prediction[0])
```
